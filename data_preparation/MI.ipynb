```python
import os
import shutil

# Reorganizar matrices por clase (T0, T1, T2)
print("Reorganizando matrices de conectividad por clase...")

# Procesar Training y Test
for dataset in ['Training', 'Test']:
    print(f"\nProcesando {dataset}...")
    
    # Procesar cada configuración de canales
    for channels in ['8', '16', '32', '64']:
        path = f"ConnectivityMatrices/{dataset}/{channels}"
        
        if not os.path.exists(path):
            print(f"  Carpeta no encontrada: {path}")
            continue
        
        print(f"  - {channels} canales")
        
        # Crear carpetas para cada clase
        for clase in ['T0', 'T1', 'T2']:
            os.makedirs(f"{path}/{clase}", exist_ok=True)
        
        # Recorrer carpetas de sujetos
        for carpeta in os.listdir(path):
            carpeta_path = f"{path}/{carpeta}"
            
            # Solo procesar carpetas de sujetos (no las de clases)
            if os.path.isdir(carpeta_path) and carpeta not in ['T0', 'T1', 'T2']:
                
                # Mover cada archivo CSV a su carpeta de clase correspondiente
                for archivo in os.listdir(carpeta_path):
                    if archivo.endswith('.csv'):
                        archivo_origen = f"{carpeta_path}/{archivo}"
                        
                        # Identificar a qué clase pertenece según el nombre
                        if '_T0_' in archivo:
                            archivo_destino = f"{path}/T0/{archivo}"
                            shutil.move(archivo_origen, archivo_destino)
                        elif '_T1_' in archivo:
                            archivo_destino = f"{path}/T1/{archivo}"
                            shutil.move(archivo_origen, archivo_destino)
                        elif '_T2_' in archivo:
                            archivo_destino = f"{path}/T2/{archivo}"
                            shutil.move(archivo_origen, archivo_destino)
                
                # Eliminar carpeta vacía del sujeto
                try:
                    os.rmdir(carpeta_path)
                except:
                    pass  # Si la carpeta no está vacía, la dejamos

print("\nReorganización completada.")
print("\nVerificando estructura final:")

# Mostrar cuántos archivos hay en cada carpeta
for dataset in ['Training', 'Test']:
    for channels in ['8', '16', '32', '64']:
        path = f"ConnectivityMatrices/{dataset}/{channels}"
        if os.path.exists(path):
            for clase in ['T0', 'T1', 'T2']:
                clase_path = f"{path}/{clase}"
                if os.path.exists(clase_path):
                    num_archivos = len([f for f in os.listdir(clase_path) if f.endswith('.csv')])
                    print(f"{dataset}/{channels}/{clase}: {num_archivos} archivos")
```
